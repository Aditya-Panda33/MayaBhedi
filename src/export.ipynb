# Step 6: Model Export for Production / Frontend Integration

Serializes the trained Wav2Vec2 weights, configuration dictionary, class mapping, and preprocessing hyperparameters into a `.pth` model package (`best_wav2vec2_deepfake_model.pth`).

```python
import os
from pathlib import Path
import torch

BASE_DIR = Path.cwd().parent if Path.cwd().name == "src" else Path.cwd()
MODELS_DIR = BASE_DIR / "models"
MODEL_PATH = MODELS_DIR / "best_wav2vec2_deepfake_model.pth"

checkpoint = torch.load(MODEL_PATH, map_location="cpu", weights_only=False)
print("Keys in Exported Model Bundle:")
for k, v in checkpoint.items():
    if isinstance(v, dict):
        print(f"  - {k}: dict with {len(v)} entries")
    else:
        print(f"  - {k}: {v}")

print(f"\nModel Bundle Size: {MODEL_PATH.stat().st_size / (1024*1024):.2f} MB")
print(f"Ready for backend/frontend integration at: {MODEL_PATH}")
```
